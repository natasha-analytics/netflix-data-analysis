```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("netflix_titles.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')
df['rating'] = df['rating'].fillna('Not Rated')
```

```python
df.isnull().sum()
```

```python
df['release_year'].value_counts().sort_index().plot(figsize=(10,5))
plt.title("Netflix Content Trend Over Years")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.show()
```

```python
df['listed_in'].str.split(', ').explode().value_counts().head(10)
```

```python
df['listed_in'].str.split(', ').explode().value_counts().head(10).plot(kind='bar', figsize=(8,4))
plt.title("Top 10 Genres on Netflix")
plt.show()
```

```python
df['country'].str.split(', ').explode().value_counts().head(10)
```

```python
df['country'].str.split(', ').explode().value_counts().head(10).plot(kind='bar', figsize=(8,4))
plt.title("Top 10 Content Producing Countries")
plt.show()
```

```python
df.groupby(['release_year','type']).size().unstack().plot(figsize=(10,5))
plt.title("Movies vs TV Shows Over Time")
plt.show()
```
